```python
class ShoppingCart:
  def __init__(self, customer_name, items= None):
    self.customer_name = customer_name
    if items == None:
      self.items = []
    else:
      self.items = items

  def add_item(self, price):
    if price>0:
      self.items.append(price)
      print(f"Item of TK {price} added to cart")
    else:
      print("Price must be a positive number.")
  
  def remove_item(self, price):
    if price in self.items:
      self.items.remove(price)
      print(f"Item of TK {price} removed from cart")
    else:
      print(f"Item of TK {price} not found in cart")
  
  def calculate_total(self):
    total = 0
    for price in self.items:
      total+=price
    return total

  def apply_discount(self):
    total = self.calculate_total()
    if total>=3000:
      final_amount = total - (total * 0.10)
    else:
      final_amount = total
    return final_amount

  def display_cart(self):
    print(f"Customer Name: {self.customer_name}")
    print(f"Items in cart: {self.items}")
    print(f"Number of items: {len(self.items)}")
    print(f"Total amount: TK {self.calculate_total()}")
    print(f"Final amouont after discount: TK {self.apply_discount()}")

  def clear_cart(self):
    self.items.clear()
    print("Cart cleared successfully.")

  def __add__(self,other):
    combined_name = self.customer_name+" & "+other.customer_name
    combined_items = self.items+other.items
    return ShoppingCart(combined_name,combined_items)


cart1 = ShoppingCart("Rashed")
cart2 = ShoppingCart("Rochi")

cart1.add_item(1200)
cart1.add_item(800)
cart1.add_item(1500)

cart1.remove_item(800)

cart2.add_item(2000)
cart2.add_item(700)
cart2.add_item(500)

cart2.remove_item(700)

cart1.display_cart()
cart2.display_cart()

combined_cart = cart1 + cart2

combined_cart.display_cart()
```
